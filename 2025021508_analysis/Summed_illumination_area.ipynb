<h2>Sum illuminations</h2>

```python
params = {'legend.fontsize': 13, # 20,
         'axes.labelsize': 17, # 25.2,
         'axes.titlesize': 17, # 25.2,
         'xtick.labelsize': 13, # 20,
         'ytick.labelsize': 13} # 20}
plt.rcParams.update(params)

def get_ill_area(probe, positions_in_pixels, pad=128):
    pr = np.pad(probe, ((pad,pad),(pad,pad)))
    illarea = np.zeros_like(pr)
    for j in range(len(positions_in_pixels) - 1):
        shifted_probe = shift(pr, positions_in_pixels[j], mode='constant', cval=0)
        illarea += np.abs(shifted_probe)
    return illarea

def plot_ill_area(illarea, obj_abs, overlap_ratio, lim=200):
    """
    lim: a factor for setting the xlim and ylim
    """
    sh = obj_abs.shape
    ncol=2
    nrow=1
    fig, ax = plt.subplots(nrow, ncol, figsize=(4 * 2.4, 2 *2.4))#(4*ncol, 4*nrow))
    fig.set_tight_layout({"w_pad":2, "h_pad":0.1, "pad":0.2, "rect":  (0.00, 0, 1, 0.95)})
    subfig0 = ax[0].imshow((np.abs(illarea)), interpolation='None')
    #ax[0].set_title('Summed illuminated area', size=13)
    ax[0].set_title(r"$\bf{(a)}$ ", x=0)


    ill_pad_row = (obj_abs.shape[0] - illarea.shape[0]) // 2
    ill_pad_col = (obj_abs.shape[1] - illarea.shape[1]) // 2
    #illarea_pad = np.pad(illarea, pad_width=((ill_pad_row+1, ill_pad_row), (ill_pad_col+1, ill_pad_col)))
    illarea_pad = np.pad(illarea, pad_width=((ill_pad_row, ill_pad_row), (ill_pad_col, ill_pad_col)))

    print(obj_abs.shape, illarea_pad.shape)
    
    marg10 = 0#588 # sh - 2*588 = 165 px
    #subfig0 = ax[1].imshow(obj_abs+np.abs(illarea_pad)/np.max(np.abs(illarea_pad)), interpolation='None', origin='upper')
    subfig1 = ax[1].imshow(obj_abs*np.abs(illarea_pad), interpolation='None', origin='upper')
    ax[1].hlines(marg10-0.5, xmin=marg10-0.5, xmax=obj_abs.shape[0]-marg10-0.5, colors='r', lw=1)
    ax[1].hlines(obj_abs.shape[0]-marg10-0.5, xmin=marg10-0.5, xmax=obj_abs.shape[1]-marg10-0.5, colors='r', lw=1)
    ax[1].vlines(marg10-0.5, ymin=marg10-0.5, ymax=obj_abs.shape[0]-marg10-0.5, colors='r', lw=1)
    ax[1].vlines(obj_abs.shape[1]-marg10-0.5, ymin=marg10-0.5, ymax=obj_abs.shape[0]-marg10-0.5, colors='r', lw=1)
    #ax[1].set_title('Illuminated area on object', size=13)
    ax[1].set_title(r"$\bf{(b)}$ ", x=0)
    
    plt.suptitle(f'Overlap {overlap_ratio}', fontsize=17, y=0.975)
    #ax[1].set_xlim(sh[1]//2 - lim, sh[1]//2 + lim)
    #ax[1].set_ylim(sh[0]//2 + lim, sh[0]//2 - lim)
    ax[1].legend(['crop view'], fontsize=10)
    
    #plt.colorbar(subfig0, ax=ax[0], fraction=0.046, pad=0.04, label='[1]')
    scalebar0 = AnchoredSizeBar(ax[0].transData, 1.0e-06 / LD.pr1_psize, r"1 $\mathbf{\mu}$m", loc='lower right', frameon=False, pad=0.3, borderpad=0.5, sep=2, label_top=True, size_vertical=1, fontproperties={'weight': 'bold'}, color='white')
    scalebar1 = AnchoredSizeBar(ax[1].transData, 1.0e-06 / LD.pr1_psize, r"1 $\mathbf{\mu}$m", loc='lower right', frameon=False, pad=0.3, borderpad=0.5, sep=2, label_top=True, size_vertical=1, fontproperties={'weight': 'bold'}, color='white')
    ax[0].add_artist(scalebar0)
    ax[1].add_artist(scalebar1)
    
    for a in ax.flatten():
        a.set_xticks([])
        a.set_yticks([])
        
    # cbar0=plt.colorbar(subfig0, ax=ax[0], fraction=0.046, pad=0.04)
    # cbar1=plt.colorbar(subfig1, ax=ax[1], fraction=0.046, pad=0.04)
    # cbar0.set_label('[1]', labelpad=0)
    # cbar1.set_label('[1]', labelpad=0)
    plt.show()
    
    return fig

```
