```python
# Plot Initial obj

savefigs = True
params = {'legend.fontsize': 13, # 20,
         'axes.labelsize': 17, # 25.2,
         'axes.titlesize': 17, # 25.2,
         'xtick.labelsize': 13, # 20,
         'ytick.labelsize': 13} # 20}
plt.rcParams.update(params)

fig14, axs = plt.subplots(nrows=1, ncols=2, figsize=(4 * 2.9, 2 *2.4))#(4 * 2.5, 4 *1.17))
#fig14.subplots_adjust(top=0.95,  bottom=0.07, wspace=0.60)
fig14.set_tight_layout({"w_pad":2, "h_pad":0.1, "pad":0.2, "rect":  (0.02, 0, 1, 1)})
subfig32 = axs[0].imshow((np.abs(obj_init)), interpolation='None', vmin=0, vmax=1, cmap='viridis')
subfig33 = axs[1].imshow(np.angle(obj_init), interpolation='None', vmin=-np.pi, vmax=0, cmap=makecmap([hsv_R, hsv_Y, hsv_G, hsv_C]))

plt.sca(axs[0])
plt.title(r"$\bf{(a)}$ ", x=0)#loc='left')
plt.sca(axs[1])
plt.title(r"$\bf{(b)}$ ", x=0)#loc='left')
cbar0=plt.colorbar(subfig32, ax=axs[0], fraction=0.046, pad=0.04)
cbar1=plt.colorbar(subfig33, ax=axs[1], fraction=0.046, pad=0.04)
cbar0.set_label('[1]', labelpad=0)
cbar1.set_label('[rad]', labelpad=0)
scalebar0 = AnchoredSizeBar(axs[0].transData, 1.0e-06 / pr1_psize, r"1 $\mathbf{\mu}$m", loc='lower right', frameon=False, pad=0.3, borderpad=0.5, sep=2, label_top=True, size_vertical=1, fontproperties={'weight': 'bold', 'size':13}, color='black')
scalebar1 = AnchoredSizeBar(axs[1].transData, 1.0e-06 / pr1_psize, r"1 $\mathbf{\mu}$m", loc='lower right', frameon=False, pad=0.3, borderpad=0.5, sep=2, label_top=True, size_vertical=1, fontproperties={'weight': 'bold', 'size':13}, color='white')
scalebar1.patch.set_alpha(0.75)
axs[0].add_artist(scalebar0)
axs[1].add_artist(scalebar1)
for ax in axs.flatten():
    ax.set_xticks([])
    ax.set_yticks([])
plt.show()
if savefigs:
    fig14.savefig(fname2.rsplit('/', 4)[0] + f'/obj_init.png', dpi=400)

```
